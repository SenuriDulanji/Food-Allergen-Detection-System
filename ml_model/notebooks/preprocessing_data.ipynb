### Categorical Columns Encoding (In-Place)

We will now encode the categorical columns in-place to keep the DataFrame clean and simple:
- **Binary columns** (`gender`, `personal_allergy_history`, `lactose_intolerance`) -> encoded to `0` / `1` in-place.
- **Ordinal columns** (`allergy_awareness`, `outside_food_frequency`) -> encoded to integer scales in-place.
- **One-Hot columns** (`province`, `blood_type`, `dietary_pattern`) -> replaced by one-hot columns.
- **Multi-Label columns** (`work_environment`, `medical_conditions`) -> replaced by binary indicators for each option.

```python
# List of categorical columns to encode
categorical_cols = [
    'gender',
    'province',
    'blood_type',
    'work_environment',
    'allergy_awareness',
    'personal_allergy_history',
    'dietary_pattern',
    'outside_food_frequency',
    'medical_conditions',
    'lactose_intolerance'
]

# 1. Clean categorical columns: strip spaces and convert special characters
df['gender'] = df['gender'].str.strip()
df['province'] = df['province'].str.strip()
df['blood_type'] = df['blood_type'].str.strip()
df['allergy_awareness'] = df['allergy_awareness'].str.strip()
df['personal_allergy_history'] = df['personal_allergy_history'].str.strip()
df['dietary_pattern'] = df['dietary_pattern'].str.rstrip(',').str.strip()
df['outside_food_frequency'] = df['outside_food_frequency'].str.replace('\u2013', '-').str.strip()
df['lactose_intolerance'] = df['lactose_intolerance'].str.strip()

# 2. In-Place Binary Encoding
# gender: Male -> 1, Female -> 0
df['gender'] = df['gender'].map({'Male': 1, 'Female': 0})

# personal_allergy_history: Yes -> 1, No -> 0
df['personal_allergy_history'] = df['personal_allergy_history'].map({'Yes': 1, 'No': 0})

# lactose_intolerance: Yes -> 1, No -> 0
df['lactose_intolerance'] = df['lactose_intolerance'].map({'Yes': 1, 'No': 0})

# 3. In-Place Ordinal Encoding
# allergy_awareness mapping
awareness_map = {
    "Not very familiar (I have heard of it but don't know much)": 0,
    "Somewhat familiar (I know the basics)": 1,
    "Very familiar (I have researched it or have direct experience)": 2
}
df['allergy_awareness'] = df['allergy_awareness'].map(awareness_map)

# outside_food_frequency mapping
freq_map = {
    "Rarely (a few times a month)": 0,
    "1-2 times a week": 1,
    "3-4 times a week": 2,
    "Daily": 3
}
df['outside_food_frequency'] = df['outside_food_frequency'].map(freq_map)

# 4. One-Hot Encoding (province, blood_type, dietary_pattern) - drops original columns automatically
one_hot_cols = ['province', 'blood_type', 'dietary_pattern']
df = pd.get_dummies(df, columns=one_hot_cols, prefix=one_hot_cols, dtype=int)

# 5. Multi-Label Encoding for work_environment
all_work_envs = set()
for val in df['work_environment'].dropna():
    parts = [p.strip() for p in val.split(',')]
    all_work_envs.update(parts)

for env in sorted(all_work_envs):
    clean_env_name = env.lower().replace(' ', '_').replace('/', '_').replace('(', '').replace(')', '').replace('-', '_')
    col_name = f"work_env_{clean_env_name}"
    df[col_name] = df['work_environment'].apply(
        lambda x: 1 if pd.notna(x) and env in [p.strip() for p in x.split(',')] else 0
    )
df = df.drop(columns=['work_environment'])

# 6. Multi-Label Encoding for medical_conditions
all_med_conds = set()
for val in df['medical_conditions'].dropna():
    parts = [p.strip() for p in val.split(',')]
    all_med_conds.update(parts)

for cond in sorted(all_med_conds):
    clean_cond_name = cond.lower().replace(' ', '_')
    col_name = f"med_cond_{clean_cond_name}"
    df[col_name] = df['medical_conditions'].apply(
        lambda x: 1 if pd.notna(x) and cond in [p.strip() for p in x.split(',')] else 0
    )
df = df.drop(columns=['medical_conditions'])

print("Processed DataFrame Shape:", df.shape)
df.head()
```

Family History Column Processing

```python
import pandas as pd
import numpy as np

# 1. Standardize the text
# Fill empty values (NaN) with an empty string, convert to lowercase, and remove extra spaces
df['family_history_clean'] = df['family_history'].fillna('').astype(str).str.lower().str.strip()

# 2. Create a general "Has Family History" flag (1 = Yes, 0 = No/Empty)
# This ignores common negative responses
def check_history(text):
    if text in ['', 'no', 'none', 'n/a', 'nothing']:
        return 0
    return 1

df['family_has_history'] = df['family_history_clean'].apply(check_history)

# 3. Extract specific conditions
# na=False prevents errors if there are still any nulls slipping through
df['family_asthma'] = df['family_history_clean'].str.contains('food allergy', regex=True, na=False).astype(int)
df['family_asthma'] = df['family_history_clean'].str.contains('asthma', regex=True, na=False).astype(int)
df['family_eczema'] = df['family_history_clean'].str.contains('eczema', regex=True, na=False).astype(int)

# 4. Define keyword maps for food allergies
# Add any other local Sri Lankan foods you expect to see in the text here
allergen_keywords = {
    'beef': ['beef'],
    'pork': ['pork'],
    'sausage': ['sausage'],
    'red_meat': ['meat', 'beef', 'pork', 'mutton'], # Grouping similar risks
    'seafood': ['seafood', 'fish', 'prawn', 'crab', 'cuttlefish / squid', 'dallo', 'shrimp', 'calms'],
    'dairy': ['dairy', 'milk', 'cheese', 'butter','curd'],
    'nuts': ['nut', 'peanut', 'cashew', 'almond'],
    'egg': ['egg'],
    'tomato': ['tomato'],
    'pineapple': ['pineapple'],
    'avocado': ['avocado'],
    'breadfruit': ['breadfruit'],
    'flour': ['flour'],
    
    
}

# 5. Loop through the dictionary and create a new 1/0 column for each food group
for feature_name, keywords in allergen_keywords.items():
    # Join keywords with a pipe (|) to create a regex "OR" pattern (e.g., 'seafood|fish|prawn')
    pattern = '|'.join(keywords)
    
    # Create the column (e.g., 'family_allergy_beef')
    df[f'family_allergy_{feature_name}'] = df['family_history_clean'].str.contains(pattern, regex=True, na=False).astype(int)

# 6. Drop the intermediate cleaning column to keep your dataset tidy
df = df.drop(columns=['family_history_clean','family_history'])
```

```python
import pandas as pd
import numpy as np

# Assuming you renamed the columns to 'native_foods' and 'other_foods'
# 1. Combine the two columns into one big text block per user
df['combined_allergies'] = df['native_food_allergy'].fillna('') + " " + df['other_food_allergy'].fillna('')

# 2. Clean the combined text
df['combined_allergies_clean'] = df['combined_allergies'].astype(str).str.lower().str.strip()

# 3. Define keyword maps (Including Sinhala/English variations)
native_allergen_keywords = {
    'jackfruit': ['jackfruit', 'kos', 'kiri kos', 'polos', 'waraka'],
    'cassava': ['cassava', 'manioc', 'tapioca'],
    'moringa': ['moringa', 'murunga'],
    'pineapple': ['pineapple', 'annasi'],
    'tomato': ['tomato', 'thakkali'],
    'rambutan': ['rambutan'],
    'sarana': ['sarana'],
    'avocado': ['avocado'],
    'breadfruit': ['breadfruit'],
    'calms': ['calms'],
    'gotukola': ['gotukola'],
    
}

# 4. Loop through the dictionary and create binary 1/0 columns
for feature_name, keywords in native_allergen_keywords.items():
    pattern = '|'.join(keywords)
    df[f'{feature_name}'] = df['combined_allergies_clean'].str.contains(pattern, regex=True, na=False).astype(int)

# 5. Drop the intermediate columns to keep the dataset clean
df = df.drop(columns=['combined_allergies', 'combined_allergies_clean','native_food_allergy','other_food_allergy'])
```

### Food Allergen Extraction & In-Place Encoding

We will now extract specific food items and spices from the free-text columns (`family_history`, `allergy_spices`, `native_food_allergy`, `other_food_allergy`) and encode all food columns in-place (overwriting the raw names like `chocolate` directly with their encoded integer scales):
- `No reaction` -> 0
- `Mild Intolerance (Stomach pain, Nausea)` -> 1
- `Mild Allergic Reaction (Skin rash, Itching, Runny nose)` -> 2
- `Severe Allergic Reaction (Swelling, Breathing difficulty)` -> 3

```python
import pandas as pd
import numpy as np

# ---------------------------------------------------------
# PART 1: Process the Google Forms Grid (from the image)
# ---------------------------------------------------------

# Create a mapping dictionary for the exact text in your Google Form
# Note: Ensure the text here exactly matches the column headers/responses in your CSV export
risk_mapping = {
    'No reaction': 0,
    'Mild Intolerance (Stomach pain, Nausea)': 1, # Adjust text if necessary
    'Mild Allergic Reaction (Skin rash, Itching, Runny nose)': 2,
    'Severe Allergic Reaction (Swelling, Breathing difficulty)': 3
}

# List the columns that correspond to the grid in the image
grid_columns = [
    'milk', 'eggs', 'peanuts', 'tree_nuts', 'wheat', 'soy', 'sesame', 'prawns',
    'crab', 'cuttlefish / squid', 'dry_fish', 'tuna', 'beef', 'pork', 'mutton', 'coconut',
    'spicy_oily'
    # Add the rest of the foods from your grid here...
]

# Apply the mapping to each column in the grid
for col in grid_columns:
    # Map the text to numbers. If a row is blank (NaN), fill it with 0 (No reaction)
    df[col] = df[col].map(risk_mapping).fillna(0).astype(int)


# ---------------------------------------------------------
# PART 2: Update the "Other" Foods to Risk Level 2
# ---------------------------------------------------------

# These are the columns we created in the previous steps (Native foods, Spices, etc.)
# They are currently formatted as 1 (Allergic) or 0 (Not Allergic)
other_allergy_columns = [
    'jackfruit',
    'cassava',
    'moringa',
    'pineapple',
    'tomato' ,
    'rambutan',
    'sarana',
    'avocado',
    'breadfruit',
    'calms',
    'gotukola',
     'mustard',
    'fenugreek',
    'cumin'
    # Add any other binary columns you want to scale up to Level 2
]

# Since they are currently 1s and 0s, multiplying the column by 2 turns the 1s into 2s!
# 0 * 2 = 0 (Remains 'No reaction')
# 1 * 2 = 2 (Becomes 'Mild Allergic Reaction')
for col in other_allergy_columns:
    df[col] = df[col] * 2

# (Family history columns remain untouched as 0s and 1s, exactly as you requested)
```
