```python
import pandas as pd

def analyze_food_risks(df, food_columns):
    """
    Analyzes food columns and returns a DataFrame ranked by overall risk and severity.
    Assumes a 0-3 ordinal scale:
    0 = No reaction, 1 = Intolerance, 2 = Mild Allergy, 3 = Severe Allergy.
    """
    risk_data = []

    for food in food_columns:
        # Calculate the Total Risk Score (sum of all 1s, 2s, and 3s)
        total_risk_score = df[food].sum()

        # Count the specific reaction severities
        severe_count = (df[food] == 3).sum()
        mild_allergy_count = (df[food] == 2).sum()
        intolerance_count = (df[food] == 1).sum()

        # Total number of people who had ANY reaction (> 0)
        total_people_reacted = (df[food] > 0).sum()

        # Append the calculated stats to our list
        risk_data.append({
            'Food / Allergen': food.replace('_', ' ').title(),
            'Total Risk Score': total_risk_score,
            'Severe Reactions (Level 3)': severe_count,
            'Mild Allergies (Level 2)': mild_allergy_count,
            'Intolerances (Level 1)': intolerance_count,
            'Total People Reacted': total_people_reacted
        })

    # Convert the results into a Pandas DataFrame
    risk_df = pd.DataFrame(risk_data)

    # Sort the data mathematically:
    # First by Total Risk Score (highest to lowest),
    # then by the number of Severe Reactions as a tie-breaker
    risk_df = risk_df.sort_values(
        by=['Total Risk Score', 'Severe Reactions (Level 3)'],
        ascending=[False, False]
    )

    return risk_df.reset_index(drop=True)
```

```python
import matplotlib.pyplot as plt

# 1. Your explicitly defined list of identified allergens
identified_allergens = [
    'milk', 'eggs', 'peanuts', 'tree_nuts', 'wheat', 'soy', 'sesame', 'prawns',
    'crab', 'cuttlefish / squid', 'dry_fish', 'tuna', 'beef', 'pork', 'mutton', 'coconut',
    'spicy_oily', 'jackfruit', 'cassava', 'moringa', 'pineapple', 'tomato', 'rambutan',
    'sarana', 'avocado', 'breadfruit', 'calms', 'gotukola', 'mustard', 'fenugreek', 'cumin'
]

# 2. Run the risk analysis function
high_risk_ranking = analyze_food_risks(df, identified_allergens)

# 3. Take the Top 15 highest risk foods
top_15_risk = high_risk_ranking.head(15).copy()

# ---> THE FIX: Sort by the visual length of the bar (Total People Reacted) <---
top_15_risk = top_15_risk.sort_values(by=['Total People Reacted', 'Severe Reactions (Level 3)'], ascending=[False, False])

# 4. Set the index and select columns for plotting
top_15_risk.set_index('Food / Allergen', inplace=True)
severity_cols = ['Severe Reactions (Level 3)', 'Mild Allergies (Level 2)', 'Intolerances (Level 1)']
plot_data = top_15_risk[severity_cols]

# 5. Define colors
colors = ['#8B0000', '#FF8C00', '#FFD700']

# 6. Create the stacked horizontal bar chart
fig, ax = plt.subplots(figsize=(12, 8))
plot_data.plot(kind='barh', stacked=True, color=colors, ax=ax, width=0.8)

# 7. Format the chart
plt.title('Top 15 Highest Risk Allergens (By Total Number of Reactions)', fontsize=16, pad=15)
plt.xlabel('Total Number of Reactions', fontsize=12)
plt.ylabel('Identified Allergen', fontsize=12)

# Invert the Y-axis so the longest bar is at the very top
plt.gca().invert_yaxis()

# Move the legend
plt.legend(title='Reaction Severity', bbox_to_anchor=(1.05, 1), loc='upper left')

plt.tight_layout()
plt.show()
```

```python
import matplotlib.pyplot as plt
import seaborn as sns

# 1. Calculate the Spearman correlation for ALL features and ALL targets
# We use the lists you already defined: 'feature_cols' and 'identified_allergens'
global_corr = df[feature_cols + identified_allergens].corr(method='spearman')

# 2. Slice the matrix to ONLY show Features (Rows) vs Targets (Columns)
targeted_global_corr = global_corr.loc[feature_cols, identified_allergens]

# 3. Setup a massive figure size so the labels are actually readable
plt.figure(figsize=(24, 18))

# 4. Plot the heatmap WITHOUT numbers (annot=False) to prevent visual clutter
sns.heatmap(
    targeted_global_corr,
    annot=False,          # Turn off text, rely on color
    cmap='RdBu_r',        # Red = Positive Risk, Blue = Negative Risk
    center=0,
    vmin=-1, vmax=1,
    linewidths=0.5,
    cbar_kws={"shrink": 0.5, "label": "Spearman Correlation"}
)

plt.title("Global Biological Matrix: All Demographics vs. All Allergens", fontsize=20, pad=20)
plt.xlabel("Allergen Targets", fontsize=16, labelpad=15)
plt.ylabel("User Demographic & Medical Features", fontsize=16, labelpad=15)

# Rotate labels so they don't overlap
plt.xticks(rotation=45, ha='right', fontsize=10)
plt.yticks(fontsize=10)

plt.tight_layout()
plt.show()
```

```python
import pandas as pd
import numpy as np

# 1. Calculate the full Spearman correlation matrix
full_corr = df[feature_cols + identified_allergens].corr(method='spearman')

# 2. Isolate Features (Rows) vs Targets (Columns)
feature_target_corr = full_corr.loc[feature_cols, identified_allergens]

# 3. Find the MAXIMUM absolute correlation for each feature across ALL 31 targets
# (If a feature is useless for every single food, its max score will be very close to 0)
# We fill NaN values with 0 here to ensure all features from feature_cols are considered
max_correlations = feature_target_corr.abs().max(axis=1).fillna(0)

# 4. Set your Noise Threshold
# 0.15 is a standard starting point for survey data.
# Anything below 0.15 is generally considered statistical noise.
THRESHOLD = 0.15

# 5. Automatically separate the "Keepers" from the "Trash"
features_to_keep = max_correlations[max_correlations >= THRESHOLD].index.tolist()

# Recalculate all dropped features based on what was NOT kept
all_dropped_features = list(set(feature_cols) - set(features_to_keep))

print("--- FEATURE OPTIMIZATION RESULTS ---")
print(f"Original Feature Count : {len(feature_cols)}")
print(f"Optimized Feature Count: {len(features_to_keep)}")
print(f"Total Dropped Features : {len(all_dropped_features)}")
print("-" * 36)

print("\n🗑️ Examples of Dropped Features (Statistical Noise or No Variance):")
print(all_dropped_features[:10])  # Print the first 10 dropped columns to verify

print("\n✅ Features Kept for the ML Pipeline:")
print(features_to_keep)
```
